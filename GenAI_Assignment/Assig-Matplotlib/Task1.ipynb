## Line Plot

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sales=pd.read_csv("../../Libraries/Matplotlib/Chocolate Sales.csv")
```

```python
sales
```

```python
sales['Date'].value_counts()
```

```python
data=sales['Date']=pd.to_datetime(sales['Date'])
```

```python
data
```

```python
sales['Amount'].value_counts()
```

```python
sales['Amount']= (sales['Amount'].str.replace('$','').str.replace(',',''))
```

```python
sales['Amount']=pd.to_numeric(sales['Amount'])
```

```python
month_sales=sales.groupby(sales['Date'].dt.to_period('M'))['Amount'].sum()
```

```python
months=month_sales.index.astype(str)
amounts=month_sales.values
```

```python
plt.plot(months,amounts)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.show()
```
