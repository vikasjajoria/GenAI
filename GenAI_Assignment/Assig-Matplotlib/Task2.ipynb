## Scatter Plot

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales=pd.read_csv("../../Libraries/Matplotlib/Chocolate Sales.csv")
```

```python
sales.dtypes
```

```python
sales['Amount'].value_counts()
```

```python
sales['Amount']=(sales['Amount'].str.replace('$','').str.replace(',',''))
```

```python
sales['Amount']=pd.to_numeric(sales['Amount'])
```

```python
plt.scatter(sales['Amount'],sales['Boxes Shipped'])
plt.title('Relationship Between Boxes Shipped and Sales Amount')
plt.xlabel('Boxes Shipped')
plt.ylabel('Sales Amount')
plt.show()
```
