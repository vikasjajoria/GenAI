## Bar Plot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
sales=pd.read_csv("../../Libraries/Matplotlib/Chocolate Sales.csv")
```

```python
sales['Amount']=sales['Amount'].str.replace('$','').str.replace(',','')
```

```python
sales['Amount']=pd.to_numeric(sales['Amount'])
```

```python
sales['Date']=pd.to_datetime(sales['Date'], format='%d-%b-%y')
```

```python
country_sales =sales.groupby('Country')['Amount'].sum().sort_values(ascending=False)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(country_sales.index, country_sales.values, color='skyblue', edgecolor='black')
plt.title('Total Sales Amount by Country')
plt.xlabel('Country')
plt.ylabel('Total Sales Amount ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.barh(country_sales.index, country_sales.values, color='salmon', edgecolor='black')
plt.title('Total Sales Amount by Country')
plt.xlabel('Total Sales Amount ($)')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```
