## Multiple Bar Plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales=pd.read_csv("../../Libraries/Matplotlib/Chocolate Sales.csv")
```

```python
sales['Amount']=sales['Amount'].str.replace('$','').str.replace(',','')
```

```python
sales['Amount']=pd.to_numeric(sales['Amount'])
```

```python
sales['Date']=pd.to_datetime(sales['Date'], format='%d-%b-%y')
```

```python
sales['Period']= 'Q' + sales['Date'].dt.quarter.astype(str)
```

```python
grouped=sales.groupby(['Country','Period'])['Amount'].sum().unstack()
```

```python
countries=grouped.index
x=np.arange(len(countries))
width=0.25
```

```python
plt.figure(figsize=(10, 6))
plt.bar(x - width, grouped['Q1'], width, label='Q1')
plt.bar(x,         grouped['Q2'], width, label='Q2')
plt.bar(x + width, grouped['Q3'], width, label='Q3')
plt.title('Chocolate Sales by Country (Q1 vs Q2 vs Q3, 2022)')
plt.xlabel('Country')
plt.ylabel('Total Sales ($)')
plt.xticks(x, countries)
plt.legend()
plt.show()
 
 
```
