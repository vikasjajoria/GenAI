```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales=pd.read_csv("../../Libraries/Matplotlib/Chocolate Sales.csv")
```

```python
sales['Amount']=sales['Amount'].str.replace('$','').str.replace(',','')
```

```python
sales['Amount']=pd.to_numeric(sales['Amount'])
```

```python
sales['Date'] = pd.to_datetime(sales['Date'], format='%d-%b-%y')
sales['Period'] = 'Q' + sales['Date'].dt.quarter.astype(str)
```

```python
grouped = sales.groupby(['Country', 'Period'])['Amount'].sum().unstack()
```

```python
grouped.plot(kind='bar', stacked=True, figsize=(9, 6),
color=['#6F4E37', '#C08552', '#F2C078'])
plt.title('Chocolate Sales by Country, Stacked by Quarter (2022)')
plt.xlabel('Country')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=0)
plt.legend(title='Quarter')
 
plt.tight_layout()
```
