```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales=pd.read_csv("../../Libraries/Matplotlib/Chocolate Sales.csv")
```

```python
sales['Amount']=sales['Amount'].str.replace('$','').str.replace(',','')
sales['Amount']=pd.to_numeric(sales['Amount'])
```

```python
plt.figure(figsize=(9, 6))
plt.hist(sales['Amount'], bins=20, color='#C08552', edgecolor='black')
 
plt.title('Distribution of Chocolate Sales Amount')
plt.xlabel('Sales Amount')
plt.ylabel('Frequency (Number of Orders)')
 
plt.tight_layout()
```
