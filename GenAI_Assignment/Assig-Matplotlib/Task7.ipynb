```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales=pd.read_csv("../../Libraries/Matplotlib/Chocolate Sales.csv")
```

```python
sales['Amount']=(sales['Amount'].str.replace('$','').str.replace(',',''))
sales['Amount']=pd.to_numeric(sales['Amount'])
```

```python
country_sales = sales.groupby('Country')['Amount'].sum()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(country_sales, labels=country_sales.index,
        autopct='%1.1f%%', startangle=90,
        colors=['#6F4E37', '#C08552', '#F2C078', '#E8A87C', '#D8973C', '#B0413E'])
 
plt.title('Chocolate Sales Market Share by Country')
 
plt.tight_layout()
```
