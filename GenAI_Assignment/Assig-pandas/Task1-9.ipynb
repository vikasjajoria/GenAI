## Task 2: Mathematical Operations on Series

```python
import pandas as pd
```

```python

marks = pd.Series([78, 85, 90, 66, 72])
```

```python
grace_marks = marks + 5
print("After adding 5 grace marks:")
print(grace_marks)
```

```python
subtracted = marks - 2
print("After subtracting 2 marks:")
print(subtracted)
```

```python
multiplied = marks * 1.05
print("After multiplying by 1.05:")
print(multiplied)
```

```python
divided = marks / 2
print("After dividing by 2:")
print(divided)
```

## Task 3: Python Functionalities on Series

```python
print("Maximum marks:", marks.max())
print("Minimum marks:", marks.min())
print("Sum of marks:", marks.sum())
print("Mean marks:", marks.mean())
```

## Task 6: Filtering & Conditional Selection

```python
above_75 = df[df['Marks'] > 75]
print("Students who scored more than 75 marks:")
above_75
```

```python
math_students = df[df['Subject'] == 'Math']
print("Students belonging to Math:")
math_students
```

```python
above_average = df[df['Marks'] > df['Marks'].mean()]
print("Students who scored more than average marks:")
above_average
```

```python
failed_students = df[df['Marks'] < 70]
print("Students who failed (marks < 70):")
failed_students
```

## Task 7: Grouping & Basic Analysis

```python
avg_marks_per_subject = df.groupby('Subject')['Marks'].mean()
print("Average marks per subject:")
avg_marks_per_subject
```

```python
students_per_subject = df.groupby('Subject')['Marks'].count()
print("Number of students per subject:")
students_per_subject
```

```python
max_marks_per_subject = df.groupby('Subject')['Marks'].max()
print("Maximum marks per subject:")
max_marks_per_subject
```

## Task 9: Mini Use Case - Sales Data Analysis

```python
sales = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Revenue': [1200, 1500, 900, 2000, 1800]
}

sales_df = pd.DataFrame(sales)
sales_df
```

```python
total_revenue = sales_df['Revenue'].sum()
print("Total revenue:", total_revenue)
```

```python
average_revenue = sales_df['Revenue'].mean()
print("Average daily revenue:", average_revenue)
```

```python
highest_day = sales_df.loc[sales_df['Revenue'].idxmax(), 'Day']
print("Day with highest revenue:", highest_day)
```

```python
above_avg_days = sales_df[sales_df['Revenue'] > average_revenue]
print("Days where revenue is above average:")
above_avg_days
```

```python
sales_df.plot(x='Day', y='Revenue', kind='bar', title='Revenue vs Day', legend=False)
```
