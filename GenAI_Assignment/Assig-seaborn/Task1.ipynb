```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("../../Libraries/Seaborn/Chocolate Sales.csv")
```

```python
data.head(5)
```

```python
data['Amount']=data["Amount"].str.replace("$","",regex=False)
data['Amount']=data["Amount"].str.replace(",","",regex=False)
data['Amount']=pd.to_numeric(data["Amount"])
```

```python
print(data.dtypes)
```

```python
sns.relplot(data=data, x="Boxes Shipped", y="Amount", hue="Country", kind="line")
plt.show()
```

```python
sns.relplot(data=data, x="Boxes Shipped", y="Amount", hue="Country", kind="scatter")
plt.show()
```
