```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("../../Libraries/Seaborn/Chocolate Sales.csv")
```

```python
data.head(5)
```

```python
data['Amount']=data['Amount'].str.replace('$','',regex=False)
data['Amount']=data['Amount'].str.replace(',','',regex=False)
data['Amount']=pd.to_numeric(data['Amount'])
```

```python
data['Date']=pd.to_datetime(data['Date'])
```

```python
sns.lineplot(data=data, x='Date', y='Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.lineplot(data=data, x='Date', y='Amount', marker='o')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.relplot(data=data, x='Date', y='Amount', col='Country', kind='line')
plt.show()
```
