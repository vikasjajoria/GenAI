```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data=pd.read_csv("../../Libraries/Seaborn/Chocolate Sales.csv")
```

```python
data.head(5)
```

```python
data['Amount']=data['Amount'].str.replace('$','',regex=False)
data['Amount']=data['Amount'].str.replace(',','',regex=False)
data['Amount']=pd.to_numeric(data['Amount'])
```

## Histogram using sns

```python
sns.histplot(data=data, x='Amount')
plt.show()
```

## KDE plot

```python
sns.kdeplot(data=data, x='Amount')
plt.show()
```

## Rug plot

```python
sns.rugplot(data=data, x='Amount')
plt.show()
```

## KDE + Histogram

```python
sns.histplot(data=data, x='Amount', kde='True')
plt.show()
```
