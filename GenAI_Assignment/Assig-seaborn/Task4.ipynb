```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("../../Libraries/Seaborn/Chocolate Sales.csv")
```

```python
data.head(5)
```

```python
data['Amount']=data['Amount'].str.replace('$','',regex=False)
data['Amount']=data['Amount'].str.replace(',','',regex=False)
data['Amount']=pd.to_numeric(data['Amount'])
```

## Categorical plot

```python
sns.countplot(data=data, x='Product')
plt.xticks(rotation=45)
plt.show()
```

## Box plot

```python
sns.boxplot(data=data, x='Product', y='Amount')
plt.xticks(rotation=45)
plt.show()
```

## violin plot

```python
sns.violinplot(data=data, x='Product', y='Amount')
plt.xticks(rotation=45)
plt.show()
```
