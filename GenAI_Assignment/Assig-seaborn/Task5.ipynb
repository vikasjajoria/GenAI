```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("../../Libraries/Seaborn/Chocolate Sales.csv")
```

```python
data.head(5)
```

```python
data["Amount"] = data["Amount"].str.replace("$", "", regex=False)
data["Amount"] = data["Amount"].str.replace(",", "", regex=False)
data["Amount"] = pd.to_numeric(data["Amount"])
```

## Pair plot

```python
sns.pairplot(data=data, vars=['Amount', 'Boxes Shipped'], hue='Country')
plt.show()
```

## Correlation heatmap

```python
correlation = data[["Amount", "Boxes Shipped"]].corr()

print(correlation)

sns.heatmap(
    correlation,
    annot=True
)

plt.show()
```
