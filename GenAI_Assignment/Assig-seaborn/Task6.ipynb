```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("../../Libraries/Seaborn/Chocolate Sales.csv")
```

```python
data["Amount"] = data["Amount"].str.replace("$", "", regex=False)
data["Amount"] = data["Amount"].str.replace(",", "", regex=False)
data["Amount"] = pd.to_numeric(data["Amount"])
```

## Bar plot

```python
sns.barplot(data=data, x='Country', y='Amount', errorbar=None)
plt.title('Average Sales Mount by Country')
plt.show()
```

## Box plot

```python
sns.boxplot(data=data, x='Country', y='Amount')
plt.show()
```

## Violin plot

```python
sns.violinplot(data=data, x='Country', y='Amount')
plt.show()
```

## Count plot

```python
sns.countplot(data=data, x='Country')
plt.show()
```
