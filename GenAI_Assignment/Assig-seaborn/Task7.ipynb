```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("../../Libraries/Seaborn/Chocolate Sales.csv")
```

```python
data["Amount"] = data["Amount"].str.replace("$", "", regex=False)
data["Amount"] = data["Amount"].str.replace(",", "", regex=False)
data["Amount"] = pd.to_numeric(data["Amount"])
```

## regplot between two numerical columns

```python
sns.regplot(data=data, x='Boxes Shipped',y='Amount')
plt.title('Regression plot : Boxes Shipped vs Amount')
plt.show()
```

```python
sns.lmplot(data=data, x="Boxes Shipped", y="Amount", hue="Country")
plt.show()
```
