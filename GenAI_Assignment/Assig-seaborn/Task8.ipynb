```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("../../Libraries/Seaborn/Chocolate Sales.csv")
```

```python
data["Amount"] = data["Amount"].str.replace("$", "", regex=False)
data["Amount"] = data["Amount"].str.replace(",", "", regex=False)
data["Amount"] = pd.to_numeric(data["Amount"])
```

## Facrgrid

```python
g = sns.FacetGrid(data, col="Country", col_wrap=3, height=3.2)
g.map_dataframe(sns.scatterplot, x="Boxes Shipped", y="Amount")
plt.show()
```

## Multi-plot dashboard

```python
sns.relplot(data=data, x="Boxes Shipped", y="Amount", hue="Country", col="Country", col_wrap=3, kind="scatter")
sns.catplot(data=data, x="Country", y="Amount", kind="box")
sns.displot(data=data, x="Amount", col="Country", col_wrap=3, kde=True)
plt.show()
```
