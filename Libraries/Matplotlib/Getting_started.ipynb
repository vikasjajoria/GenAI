```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df['City']=df['City'].astype('category')
```

```python
data=df.groupby('Age')['CGPA'].mean().head(10)
```

```python
data
```

```python
x=data.index
y=data.values
```

```python
plt.figure(figsize=(10,5))
plt.title('CGPA Distribution as per age')
plt.xlabel('AGE')
plt.ylabel('CGPA')
plt.plot(x,y,color='red',linestyle='--',linewidth=1.5,marker='o',markersize=7)
plt.grid()
plt.show()
```
