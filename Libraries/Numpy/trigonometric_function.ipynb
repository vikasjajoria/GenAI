## Stacking and Splitting

```python
array_1=np.arange(1,13).reshape(3,4)
```

```python
array_1
```

```python
array_2=np.arange(13,25).reshape(3,4)
```

```python
array_2
```

```python
horizantal_array=np.hstack((array_1,array_2))
```

```python
horizantal_array
```

```python
vertical_array=np.vstack((array_1,array_2))
```

```python
vertical_array
```

```python
np.hsplit(vertical_array,2)
```

```python
np.vsplit(vertical_array,2)
```
