```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
sales=pd.read_csv("Student_Depression_Dataset.csv")
```

```python
sales.head(5)
```

```python
sns.lineplot(data=sales,x='Age',y='CGPA',hue='Gender',style='Dietary Habits')
plt.show()
```

```python
sns.relplot(data=sales,x='Age',y='CGPA',kind='line',hue='Gender')
plt.show()
```

```python
tips=sns.load_dataset('tips')
```

```python
tips.head()
```

```python
sns.lineplot(data=tips,x='day',y='total_bill')
plt.show()
```

```python
sns.relplot(data=tips,x='day',y='total_bill',kind='line',hue='smoker',style='sex')
```

```python
sns.lineplot(data=tips,x='day',y='total_bill',hue='smoker',style='time')
plt.show()
```
