```python
def fam_type(num):
    if num>=1 and num<=3:
        return 'Small Family'
    elif num>=3 and num<=5:
        return 'Medium Size Family'
    else:
        return 'Large Family'
```

```python
df['Family Type']=df['Family Size'].apply(fam_type)
```
