```python
def even_or_odd_instance(num):
    if isinstance(num, int):
        if num%2==0:
            print(f"{num} is even")
        else:
            print(f"{num} is odd")
    elif isinstance(num, str):
        print("Invalid Datatype")
    else:
        print(f"{num} is an invalid value for this function")            
```

```python
even_or_odd_instance(True)
```

```python
even_or_odd_instance(41.15)
```

```python
even_or_odd_instance(88254)
```

## Recursive function

by a recursive function we can say that when a function call itself again and again

```python
def factorial(num):
    if num == 0 or num == 1:
        print(f"The factorial of {num} is: 1")
    else:
        factorial = 1
        for i in range(1, num + 1):
            factorial = factorial * i
        print(f"The factorial of {num} is: {factorial}")  
```

```python
factorial(5)
```
