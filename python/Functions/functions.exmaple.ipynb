###define a function that would behave as a password strength chaecker

```python
def password_strength(password):
    if len(password)<8:
        print(f"Password must be atleast 8 characters long")
    elif not any(char.islower() for char in password):
        print(f"Password must contain at least one lowercase char")
    elif not any(char.isupper() for char in password):
        print(f"Password must contain at least one Uppercase char")
    elif not any(char.isdigit() for char in password):
        print(f"Password must contain at least one number")
    elif not any(char in "!@#" for char in password):
        print(f"Password must contain at least one special char")
    else:
        return "Strong Password"        
```

```python
print(password_strength("helloA!123"))
```

```python
print(password_strength("hello!123"))
```

Palindrome

```python
def palindrome(string):
    string_value=string.lower().replace(" ", "")
    if string_value==string_value[::-1]:
        return "This current string a palindrome"
    else:
        return "This current string not a palindrome"    
```

```python
palindrome("racecar")
```

```python
palindrome("python")
```

```python
palindrome("LEvEl")
```

```python
palindrome("a man a plan a canal panama")
```
