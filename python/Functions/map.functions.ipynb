##squaring all items of a list

```python
def square(list_items):
    empty_list=[]

    for item in list_items:
      squared_item=item**2
      empty_list.append(squared_item)
    print(empty_list)
```

```python
square([25,7,5,8,88,5])
```

using map

```python
print(list(map(lambda x:x**2, [1,5,55,65,24])))
```
