```python
class Cartoon:

    def __init__(self,name, show, catchphrase):
        self.name = name
        self.show = show
        self.catchphrase = catchphrase

    def introduction(self):
        print(f"Hello my name is {self.name}")

    def speak(self):
        print(f"{self.name} says {self.catchphrase}")
                
```

```python
Micky_mouse = Cartoon("Micky Mouse", "Micky mouse clubhouse", "oh boy !")
```

```python
Micky_mouse.show
```

```python
Micky_mouse.introduction()
```

```python
Micky_mouse.speak()
```

```python
Spongebob = Cartoon("Spongbob squarepants","Spongbob squarepants","I am ready")
```

```python
Spongebob.speak()
```
