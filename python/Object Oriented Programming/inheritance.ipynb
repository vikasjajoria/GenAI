```python
class Avenger:

    def __init__(self,name,power):
        self.name=name
        self.power=power

    def introduction(self):
        print(f"Hello, My name is {self.name}")  


class Ironman(Avenger):

    def __init__(self,name,power,weapon):
        super().__init__(name,power)
        self.weapon=weapon

    def weapon_details(self):
        print(f"I use  {self.weapon}")
```

```python
iron=Ironman("Ironmen","Genious intellect","Repulsor beams")
```

```python
iron.weapon
```

```python
iron.weapon_details()
```

```python
iron.introduction()
```

Multilevel Inheritancemmm


```python
class Grandparent:

    def __init__(self,nationality, baldness, anger_issues):
        self.nationality=nationality
        self.baldness=baldness
        self.anger_issues=anger_issues

    def introduction(self):
        print(f"Hello my nationality is : {self.nationality} and about my anger_issues if you want to ask about then the answer is {self.anger_issues}")    
```

```python
class Parent(Grandparent):

    def __init__(self,nationality,baldness,anger_issues,intelligence):
        super().__init__(nationality,baldness,anger_issues)
        self.intelligence=intelligence

    def intelligence_details(self):
        print(f"About my intelligenceif you guys want to know that wheather if i am intelligence or not then the answer is : {self.intelligence}")    
```

```python
class Child(Parent):

    def __init__(self,nationality,baldness,anger_issues,intelligence,charm):
        super().__init__(nationality,baldness,anger_issues,intelligence)
        self.charm=charm

    def char_details(self):
        print(f"About my charm if you guys to know that whether charming or not : {self.charm}")    
```

```python
parent_1=Parent("Indian","Yes","No","Yes")
```

```python
parent_1.introduction()
```

```python
child_1=Child("Inadian","No","Yes","Yes","Yes")
```

```python
child_1.introduction()
```

```python
child_1.char_details()
```

```python
child_1.nationality="Cristan"
```

```python
child_1.introduction()
```

```python
class Ironman:

    def __init__(self,name,power,suit,weapon,nationality="Amerincan"):
        self.name=name
        self.power=power
        self.suit=suit
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"My name is : {self.name} and power are : {self.power}")
        

class Captionamerica:

    def __init__(self,name,power,suit,weapon,nationality="American"):
        self.name=name
        self.power=power
        self.suit=suit
        self.weapon=weapon
        self.nationality=nationality 

    def weapon_details(self):
        print(f"I carry the weapon is : {self.weapon}")         
```

```python
class Hulk(Ironman,Captionamerica):

    def __init__(self,name,power,suit,weapon,anger_issues,nationality="American"):
        super().__init__(name,power,suit,weapon,nationality)
        self.anger_issues=anger_issues

    def anger_description(self):
        print(f"About my anger issues if i am having or not then the answer is : {self.anger_issues}")    
```

```python
hulk_guy=Hulk("Hulk","Amazing power and strength","Mark 85 ironman suit","Repulsor beams and sheild","Yes","American")
```

```python
hulk_guy.introduction()
```

```python
hulk_guy.weapon_details()
```

```python
hulk_guy.anger_description()
```
