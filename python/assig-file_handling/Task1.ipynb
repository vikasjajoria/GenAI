File Handling

1) write each sale on a new line

```python
sales =[1200,450,980,1500,3000]
```

```python
file = open("sales_data.txt","w")
```

```python
for sale in sales:
    file.write(str(sale)+"\n")

file.close()
```

Reopen and print the content

```python
file = open("sales_data.txt","r")
```

```python
print(file.read())
```

```python
file.close()
```

Write the data inn comma seprated

```python
sales = [1200,450,980,1500,3000]
```

```python
file = open("sales_data1.txt", "w")
```

```python
file.write(",".join(map(str,sales)))
```

```python
file.close()
```
