Read file in different ways

1) Read the entire file

```python
file = open("sales_data.txt","r")

data = file.read()

print(data)
```

```python
file.close()
```

2) Read the file line by line

```python
file = open("sales_data.txt","r")

line1 =file.readline()
line2 =file.readline()
line3 =file.readline()
line4 =file.readline()
line5 =file.readline()
```

```python
print(line1)
print(line2)
print(line3)
print(line4)
print(line5)
```

3) Read all lines and convert into a list

```python
file =open("sales_data.txt","r")

lines = file.readlines()

sales = []

for line in lines:
    sales.append(int(line.strip()))

print(sales)
     
```

```python
file.close()
```
