Append new sales

```python
file = open("sales_data.txt","a")

file.write("5000\n")
file.write("2500\n")
file.write("1700\n")
```

```python
file.close()
```

Reopen the file and print it

```python
file = open("sales_data.txt","r")

data = file.read()

print(data)
```

```python
file.close()
```

Count the total number of lines

```python
file = open("sales_data.txt","r")

lines = file.readlines()

print("Total number of lines:", len(lines))
```

```python
file.close()
```
