Generate summary report from file

```python
file = open("sales_data.txt","r")

lines = file.readlines()

file.close()

```

Convert lines into intergers

```python
sales = []

for line in lines:
    sales.append(int(line.strip()))

print(sales)    
```

Calculate result

```python
total_sale=sum(sales)
highest_sale=max(sales)
lowest_sale=min(sales)
average_sale=total_sale/len(sales)
```

```python
print("Total sale:", total_sale)
print("Highest sale:", highest_sale)
print("Lowest sale:", lowest_sale)
print("averahe sale:", average_sale)
```
